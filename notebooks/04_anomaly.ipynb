```python
import pandas as pd 
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
import joblib
import os
print("ready!")
```

```python
df = pd.read_csv('../data/processed/weather_clean.csv')
print(df.shape)
```

```python
features = ['avg_temp_c', 'min_temp_c', 'max_temp_c', 'precipitation_mm', 'avg_wind_speed_kmh']
x = df[features].dropna()
print("shape:", x.shape)
```

```python
x_sample = x.sample(n=100000, random_state=42)
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x_sample)
print("scaled shape:", x_scaled.shape)
```

```python
iso_model = IsolationForest(
    n_estimators=100,
    contamination=0.05,
    random_state=42
)

iso_model.fit(x_scaled)
print("isolation forest trained!")
```

```python
preds = iso_model.predict(x_scaled)
scores = iso_model.decision_function(x_scaled)
normal = (preds == 1).sum()
extreme = (preds == -1).sum()
print("normal days:", normal)
print("extreme days:", extreme)
print("extreme %:", round(extreme/len(preds)*100, 2))
```

```python
os.makedirs('../models/anomaly', exist_ok=True)
joblib.dump(iso_model, '../models/anomaly/model.pkl')
joblib.dump(scaler, '../models/anomaly/scaler.pkl')
print("anomaly model saved!")
```
