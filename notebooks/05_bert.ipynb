```python
import json
import os
print("no libraries needed!")
```

```python
def classify_intent(text):
    text = text.lower()

    rain_words = ['rain', 'rainfall', 'umbrella', 'wet', 'drizzle', 'shower']
    forecast_words = ['forecast', 'week', 'tomorrow', 'next', '7 days', 'days']
    extreme_words = ['storm', 'hurricane', 'flood', 'extreme', 'alert', 'warning']
    temp_words = ['temperature', 'hot', 'cold', 'warm', 'heat', 'freeze', 'temp']
    activity_words = ['outside', 'outdoor', 'walk', 'run', 'picnic', 'go out']

    if any(w in text for w in rain_words):
        return 'rain_prediction'
    elif any(w in text for w in extreme_words):
        return 'extreme_weather'
    elif any(w in text for w in forecast_words):
        return 'forecast'
    elif any(w in text for w in temp_words):
        return 'temperature'
    elif any(w in text for w in activity_words):
        return 'activity_advice'
    else:
        return 'general'

print("intent classifier ready!")
```

```python
tests = [
    "will it rain tomorrow?",
    "7 day weather forecast",
    "is there a storm coming?",
    "how hot will it be?",
    "can i go for a walk today?"
]

for t in tests:
    print(f"'{t}' -> {classify_intent(t)}")
```
